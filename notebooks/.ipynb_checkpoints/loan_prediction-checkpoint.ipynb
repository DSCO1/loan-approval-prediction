## 🏦 Loan Approval Prediction using Machine Learning

### 📖 Project Overview

This project aims to build a machine learning model that predicts whether a bank customer will subscribe to a term deposit (loan product) based on their personal and financial information.

The dataset used in this project is the Bank Marketing Dataset.

---

### 🎯 Objective

The main objective of this project is to:

- Analyze customer data using Exploratory Data Analysis (EDA)
- Identify key factors influencing loan subscription
- Build a classification model
- Evaluate model performance

---

### 📊 Dataset Description

The dataset contains 45,000+ records.

Features include:

- age  
- job  
- marital  
- education  
- housing  
- loan  
- contact  
- duration  
- campaign  
- pdays  
- previous  
- y (target variable)

---

### 🤖 Machine Learning Approach

1. Data Loading  
2. Data Cleaning  
3. EDA  
4. Encoding  
5. Model Training  
6. Evaluation  

---

### 🛠️ Technologies Used

- Python

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("../data/bank.csv", sep=';')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
sns.countplot(x='y', data=df)
plt.title("Loan Subscription Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(x='job', hue='y', data=df)
plt.xticks(rotation=45)
plt.title("Job vs Loan Subscription")
plt.show()
```

```python
sns.countplot(x='marital', hue='y', data=df)
plt.title("Marital Status vs Loan Subscription")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='education', hue='y', data=df)
plt.xticks(rotation=45)
plt.title("Education vs Loan Subscription")
plt.show()
```

```python
sns.countplot(x='housing', hue='y', data=df)
plt.title("Housing Loan vs Subscription")
plt.show()

sns.countplot(x='loan', hue='y', data=df)
plt.title("Personal Loan vs Subscription")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='y', y='duration', data=df)
plt.title("Duration vs Loan Subscription")
plt.show()
```

```python
df['y'] = df['y'].map({'yes':1, 'no':0})
```
